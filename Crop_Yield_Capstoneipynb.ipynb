```python
df = Yield[(Yield['Item'] == MY_CROP) & (Yield['Area'] == MY_COUNTRY)].copy()

print(f'{len(df)} tracks in {MY_CROP} for {MY_COUNTRY}')
df[['Area', 'pesticides_tonnes', 'hg/ha_yield']].head()
```

```python

df_obs = df.drop(columns=['Area', 'Item', 'Unnamed: 0', 'index'], errors='ignore')

df_obs = df_obs.sort_values('Year')

print("=== OBSERVATION 1: YIELD TREND ===")
first_year = df_obs.iloc[0]
last_year = df_obs.iloc[-1]
print(f"Start Year: {first_year['Year']} | Yield: {first_year['hg/ha_yield']:.0f} hg/ha")
print(f"End Year:   {last_year['Year']} | Yield: {last_year['hg/ha_yield']:.0f} hg/ha")
print(f"Average Yield: {df_obs['hg/ha_yield'].mean():.0f} hg/ha")

print("\n=== OBSERVATION 2: CLIMATE ===")
print(f"Rainfall vs Yield correlation: {df_obs['average_rain_fall_mm_per_year'].corr(df_obs['hg/ha_yield']):.3f}")
print(f"Temperature vs Yield correlation: {df_obs['avg_temp'].corr(df_obs['hg/ha_yield']):.3f}")

print("\n=== OBSERVATION 3: PESTICIDES ===")
print(f"Pesticides vs Yield correlation: {df_obs['pesticides_tonnes'].corr(df_obs['hg/ha_yield']):.3f}")
```

**Observation 1 (The Yield Trend)**:

Between 1990 and 2013, Ghana's maize yield increased significantly, rising from 11,889 hg/ha to 17,240 hg/ha, with an average yield of 15,399 hg/ha over the 23-year period. This upward trend suggests that agricultural interventions, improved seed varieties, or better farming practices have successfully boosted maize production in Ghana over the last two decades.

**Observation 2 (The Climate Impact)**:

Interestingly, the average temperature has a strong positive correlation with yield (0.607). However, we discovered that the rainfall data for Ghana in this dataset is a constant value (or missing), resulting in a 'NaN' correlation. This means we cannot use rainfall to predict yield in this specific dataset, and future models must rely on temperature and time.

**Observation 3 (The Pesticides)**:

Pesticide use shows a weak positive correlation with yield (0.250). While the relationship is positive—meaning years with more pesticide use tended to have slightly higher yields—the effect is not strong. This suggests that while pesticides help protect the crop, other factors (like temperature, time, or fertilizer) are playing a larger role in driving yields up.
