```python
import pandas as pd
import numpy as np
import seaborn as sn
import matplotlib.pyplot as plt
```

```python
Yield = pd.read_csv("/content/yield_df.csv")
```

```python
MY_CROP = 'Maize'
df = Yield[Yield['Item'] == MY_CROP].copy()
```

```python
df = df.drop(columns=['Unnamed: 0', 'Item'], errors='ignore')
```

```python
print(f" Data loaded for {MY_CROP}")
print(f"Total rows: {len(df)}")
print(f"Total countries (Area): {df['Area'].nunique()}")
print(df.head())
```

```python
yearly_avg = df.groupby('Year')['hg/ha_yield'].mean()
print("=== OBSERVATION 1: YIELD TREND ===")
print(f"Start Year: {yearly_avg.index.min()} | Avg Yield: {yearly_avg.iloc[0]:.0f} hg/ha")
print(f"End Year: {yearly_avg.index.max()} | Avg Yield: {yearly_avg.iloc[-1]:.0f} hg/ha")

print("\n=== OBSERVATION 2: CLIMATE ===")
print(f"Rainfall vs Yield correlation: {df['average_rain_fall_mm_per_year'].corr(df['hg/ha_yield']):.3f}")
print(f"Temperature vs Yield correlation: {df['avg_temp'].corr(df['hg/ha_yield']):.3f}")

country_avg = df.groupby('Area')['hg/ha_yield'].mean().sort_values()
print("\n=== OBSERVATION 3: COUNTRY VARIATION ===")
print(f"Lowest yielding country: {country_avg.index[0]} ({country_avg.iloc[0]:.0f} hg/ha)")
print(f"Highest yielding country: {country_avg.index[-1]} ({country_avg.iloc[-1]:.0f} hg/ha)")
```

**Observation 1 (The Global Trend):**

Between 1990 and 2013, the global average yield of Maize almost doubled, rising from 27,706 hg/ha to 47,145 hg/ha. This strong upward trend suggests that global agricultural practices, improved seed varieties, and better farming policies have successfully boosted maize production over the last two decades.

**Observation 2 (The Climate Impact):**

Both temperature and rainfall show a negative correlation with maize yield. Temperature has a moderate negative correlation (-0.545), and rainfall has a weak negative correlation (-0.179). This means that, globally, years with higher average temperatures tend to have lower maize yields. We must include these climate variables in our regression model to account for environmental stress on crops.

**Observation 3 (Country Variation):**

There is massive inequality in maize yields between countries. The lowest yielding country in the dataset is Botswana at just 2,350 hg/ha, while the highest is Qatar at 136,593 hg/ha. This means the ministry cannot use a single blanket policy for maize; interventions and support must be highly tailored to specific regions, especially those struggling like Botswana.
